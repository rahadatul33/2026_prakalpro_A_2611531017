## 2.1 Dasar Teori
Praktikum ini membahas tiga hal yang saling berkaitan dalam proses pengembangan program, yaitu dasar-dasar bahasa pemrograman Python, penggunaan Visual Studio Code sebagai editor untuk menulis dan menjalankan kode, serta peran Git dan GitHub dalam mencatat riwayat perubahan sekaligus menyimpan proyek secara daring.

1. Python<br>Python merupakan bahasa pemrograman tingkat tinggi yang dirancang dengan aturan penulisan kode yang relatif sederhana. Hal tersebut membuat Python lebih mudah dipelajari, khususnya bagi pengguna yang baru mulai memahami pemrograman. Penggunaannya cukup luas, seperti dalam pembuatan aplikasi, pengolahan data, maupun pengembangan perangkat lunak.
2. Visual Studio Code<br>Visual Studio Code atau yang biasa disebut VS Code adalah perangkat lunak yang digunakan untuk mengerjakan kode program. Melalui VS Code, kode dapat ditulis, diedit, dan dijalankan. Aplikasi ini juga dapat digunakan untuk bahasa pemrograman seperti Python dan menyediakan fasilitas pendukung berupa terminal, debugging, serta integrasi dengan Git.
3. Git<br>Git merupakan sebuah sistem pengontrol versi (version control) yang berfungsi untuk merekam setiap perubahan yang terjadi pada suatu program. Melalui Git, seorang programmer dapat menelusuri riwayat perubahan serta mengelola berbagai versi program secara terstruktur.
4. GitHub<br>Git merupakan sebuah sistem pengontrol versi (version control) yang berfungsi untuk merekam setiap perubahan yang terjadi pada suatu program. Melalui Git, seorang programmer dapat menelusuri riwayat perubahan serta mengelola berbagai versi program secara terstruktur.
5. Repository<br>Repository merupakan wadah penyimpanan bagi suatu proyek, mencakup kode program beserta berkas-berkas pendukung lainnya. Penyimpanan repository dapat dilakukan secara lokal di komputer pengguna maupun secara daring melalui GitHub.
6. Commit dan Push<br>Proses commit dilakukan untuk mencatat serta menyimpan setiap perubahan yang telah dibuat pada repository lokal. Sementara itu, proses push berfungsi untuk mengunggah perubahan tersebut dari repository lokal menuju repository yang tersimpan di GitHub.
