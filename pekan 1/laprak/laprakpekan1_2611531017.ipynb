## 2.2.2 Membuat Repository GitHup

Setelah progam di VS Code selesai dilakukan, langkah selanjutnya adalah mengintegrasikan proyek tersebut ke dalam platform GitHub. Fungsi utama repository sebagai tempat penyimpanan kode progam.

Tahapan yang dilakukan meliputi:
1. Pembuatan repository baru pada akun GitHub.
2. Pembukaan folder proyek melalui aplikasi VS Code.
3. Pengaksesan fitur terminal bawaan yang terdapat pada VS Code.
4. Pelaksanaan inisialisasi Git terhadap folder proyek yang telah dibuka.
5. Pemasukan seluruh file proyek ke dalam staging area.
6. Pembuatan commit guna mencatat setiap perubahan yang terjadi pada file.
7. Penyambungan antara repository lokal dengan repository yang sudah tersedia di GitHub.
8. Pelaksanaan proses push, yaitu mengunggah file beserta kode program menuju repository GitHub.

Perintah yang digunakan:


```python
git init
git add . 
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

**Penjelasan :**
1. git init dipakai untuk menginisialisasi repository Git yang baru di dalam folder proyek yang akan dikelola.
2. git add . perintah ini menambahkan semua file yang telah diubah atau dibuat ke staging area, sebagai langkah persiapan sebelum commit dilakukan.
3. git commit digunakan untuk mencatat dan menyimpan seluruh perubahan yang ada di staging area ke dalam riwayat repository, disertai pesan sebagai penjelasan.
4. git branch -M berguna untuk mengganti nama branch utama pada repository menjadi "main".
5. git remote add origin perintah ini menghubungkan repository yang ada di komputer (lokal) dengan repository yang tersimpan di GitHub.
6. git push dipakai untuk mengunggah seluruh commit dari repository lokal ke repository GitHub agar tersimpan secara online.
