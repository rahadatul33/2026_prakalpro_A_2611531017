## 2.1 Dasar Teori
Pada praktikum ini dibahas beberapa materi dasar dalam bahasa pemrograman Python, yaitu string, numerik, input, boolean, dan konstanta. Materi tersebut digunakan dalam membuat program sederhana dan memiliki fungsi yang berbeda sesuai dengan kebutuhan program. Dasar teori dari masing-masing materi yang dipelajari adalah sebagai berikut:

### 2.1.1 String <br>
String merupakan tipe data yang digunakan untuk menyimpan data berupa teks atau kumpulan karakter. String dapat berisi huruf, angka, simbol, maupun kalimat. Dalam Python, string biasanya ditulis menggunakan tanda petik satu (' ') atau tanda petik dua (" "). Tipe data ini sering digunakan untuk menyimpan informasi yang berupa teks, seperti nama, alamat, jenis kelamin, dan agama.

### 2.1.2 Numerik <br>
 Numerik merupakan tipe data yang digunakan untuk menyimpan nilai berupa angka. Python memiliki beberapa jenis data numerik, di antaranya int untuk bilangan bulat, float untuk bilangan desimal, dan complex untuk bilangan kompleks. Tipe data numerik dapat digunakan dalam berbagai operasi matematika seperti penjumlahan, pengurangan, perkalian, dan pembagian.

### 2.1.3 Input <br> 
Input merupakan proses memasukkan data ke dalam program dari pengguna. Pada Python, fungsi input() digunakan untuk meminta masukan dari pengguna ketika program sedang dijalankan. Dengan adanya input, program dapat menerima nilai yang berbeda sesuai dengan data yang dimasukkan oleh pengguna. Data yang diterima melalui input() pada dasarnya berupa string. Jika data tersebut akan digunakan sebagai angka dalam perhitungan, data perlu diubah terlebih dahulu menggunakan fungsi seperti int() atau float(). Dengan demikian, input dapat digunakan untuk membuat program yang lebih interaktif.

### 2.1.4 Boolean <br> 
Boolean merupakan tipe data yang digunakan untuk menyatakan nilai kebenaran suatu kondisi. Boolean hanya memiliki dua nilai, yaitu True dan False. Nilai tersebut dapat diperoleh dari hasil perbandingan maupun digunakan secara langsung dalam program. Boolean sering digunakan ketika program perlu menentukan apakah suatu kondisi benar atau salah.

### 2.1.5 Konstanta <br> 
Konstanta adalah nilai yang digunakan sebagai nilai tetap dalam program. Dalam Python terdapat beberapa konstanta bawaan, seperti True, False, dan None. Nilai True dan False berkaitan dengan tipe data boolean, sedangkan None digunakan untuk menunjukkan tidak adanya suatu nilai. Konstanta bawaan tersebut memiliki aturan khusus dan tidak dapat diberi nilai baru.

### 2.2.4 Program "Boolean_2611531017"

```python
# deklarasi variabel dengan tipe data Boolean
is_lulus_1017 = True
is_cumlaude_1017 = True

# Menggunakan Boolean
nilai_1017 = 85
batas_lulus_1017 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_1017= nilai_1017 >= batas_lulus_1017 # Hasilnya akan True

print("=== Check Kelulusan===")
print("Nilai:", nilai_1017)
print("Apakah Lulus?", status_kelulusan_1017)
if is_lulus_1017 and is_cumlaude_1017:
    print("Selamat, Anda lulus dengan predikat Cum Laude! ")
```

**Penjelasan**

2. Baris 2 berfungsi untuk mendeklarasikan variabel is_lulus_1017 dan memberinya nilai Boolean True, yang menandakan status kelulusan bernilai benar.
3. Baris 3 berfungsi untuk mendeklarasikan variabel is_cumlaude_1017 dan memberinya nilai Boolean True, yang menandakan status cum laude bernilai benar.
4. Baris 5 berfungsi sebagai komentar yang menjelaskan bahwa bagian program selanjutnya menunjukkan penggunaan tipe data Boolean dalam sebuah kondisi.
5. Baris 6 berfungsi untuk mendeklarasikan variabel nilai_1017 dan memberinya nilai integer 85, yang merepresentasikan nilai akhir mahasiswa.
6. Baris 7 berfungsi untuk mendeklarasikan variabel batas_lulus_1017 dan memberinya nilai integer 75, yang merepresentasikan batas minimal nilai kelulusan.
7. Baris 9 berfungsi sebagai komentar yang menjelaskan bahwa baris berikutnya digunakan untuk menentukan nilai Boolean berdasarkan hasil suatu kondisi.
8. Baris 10 berfungsi untuk menentukan status kelulusan dengan membandingkan nilai_1017 dan batas_lulus_1017 menggunakan operator relasional >= (lebih besar sama dengan). Karena 85 >= 75 bernilai benar, hasil perbandingan berupa Boolean True disimpan dalam variabel status_kelulusan_1017.

### 2.2.5 Program "Konstanta_2611531017"

```python
# Buat file dengan nama Konstanta_1017.py
# Progam ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1017

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_1017 = float(input('Masukan nilai jari-jari:'))
luas_1017 = PI * jari_1017 * jari_1017
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1017, luas_1017))
```

**Penjelasan**
1. Baris 1 sampai 3 adalah komentar yang berisi keterangan tentang nama file, tujuan program, dan aturan penulisan nama variabel. Baris ini hanya catatan dan tidak diproses oleh program.
2. Baris 5 berfungsi mengambil fungsi Final dari modul typing. Fungsi ini digunakan untuk menandai bahwa suatu variabel adalah konstanta, yaitu nilai yang tidak boleh diubah selama program berjalan.
3. Baris 6 berfungsi membuat variabel bernama PI dengan tipe Final dan nilai tetap 3.14. Nilai ini digunakan sebagai nilai pi dalam perhitungan lingkaran.
4. Baris 7 berfungsi menampilkan nilai PI ke layar dengan perintah print(). Tanda %f berarti nilai akan ditampilkan dalam bentuk bilangan pecahan (desimal).
5. Baris 8 berfungsi meminta pengguna memasukkan angka jari-jari lingkaran lewat perintah input(). Perintah float() digunakan agar angka yang dimasukkan pengguna, yang awalnya berbentuk teks, diubah menjadi bilangan desimal. Hasilnya disimpan dalam variabel jari_1017.
6. Baris 9 berfungsi menghitung luas lingkaran dengan rumus pi dikali jari-jari dikali jari-jari (PI × jari_1017 × jari_1017). Hasil hitungannya disimpan dalam variabel luas_1017.
7. Baris 10 berfungsi menampilkan hasil akhir ke layar, yaitu nilai jari-jari dan luas lingkaran yang sudah dihitung. Tanda %.2f berarti angka yang ditampilkan dibulatkan menjadi dua angka di belakang koma.
