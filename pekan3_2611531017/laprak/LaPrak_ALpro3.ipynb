## 2.1 Dasar Teori <br>
Pada praktikum ini dibahas beberapa jenis operator dalam bahasa pemrograman Python, yaitu operator aritmatika, assignment, bitwise, logika, keanggotaan, serta perbandingan. Operator-operator tersebut digunakan untuk mengolah data dan membentuk kondisi tertentu sesuai kebutuhan program, sehingga program dapat berjalan dan menghasilkan keluaran sesuai yang diharapkan. Dasar teori dari masing-masing operator yang dipelajari adalah sebagai berikut:

## 2.1.1 Operator Aritmatika<br>
Operator aritmatika merupakan operator yang digunakan untuk melakukan perhitungan matematis terhadap dua buah nilai atau lebih. Operator ini terdiri dari penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), sisa bagi atau modulus (%), dan pangkat (**). Pembagian bulat menghasilkan nilai hasil bagi tanpa angka desimal, sedangkan sisa bagi menghasilkan nilai sisa dari suatu pembagian. Operator aritmatika banyak digunakan dalam program yang melibatkan proses perhitungan angka.

## 2.1.2 Operator Assignment<br>
Operator assignment digunakan untuk memberikan atau memperbarui nilai suatu variabel. Bentuk paling sederhana adalah tanda sama dengan (=), yang digunakan untuk menyimpan suatu nilai ke dalam variabel. Selain itu, Python juga menyediakan operator assignment gabungan seperti penambahan (+=), pengurangan (-=), perkalian (*=), pembagian (/=), pembagian bulat (//=), sisa bagi (%=), dan pemangkatan (**=). Operator assignment gabungan memungkinkan suatu variabel diperbarui nilainya berdasarkan nilai sebelumnya tanpa perlu menuliskan ulang nama variabel secara lengkap, sehingga penulisan program menjadi lebih ringkas.

## 2.1.3 Operator Bitwise<br>
Operator bitwise merupakan operator yang bekerja pada tingkat bit atau representasi biner dari suatu bilangan. Operator ini terdiri dari AND (&), OR (|), XOR (^), NOT (~), geser kiri (<<), dan geser kanan (>>). Operator AND, OR, dan XOR membandingkan bit dari dua bilangan, operator NOT membalik seluruh bit dari suatu bilangan, sedangkan operator geser kiri dan geser kanan menggeser posisi bit sejumlah tertentu ke arah kiri atau kanan. Operator bitwise sering digunakan dalam program yang berkaitan dengan pengolahan data pada level biner.

## 2.1.4 Operator Logika<br>
Operator logika digunakan untuk mengombinasikan atau membalik nilai boolean dalam suatu program. Operator ini terdiri dari and, or, dan not. Operator and menghasilkan nilai True apabila kedua kondisi bernilai True, operator or menghasilkan nilai True apabila salah satu kondisi bernilai True, sedangkan operator not digunakan untuk membalik nilai boolean yang ada. Operator logika biasanya digunakan ketika suatu program perlu menggabungkan lebih dari satu kondisi sekaligus.

## 2.1.5 Operator Keanggotaan<br>
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu kumpulan data seperti list, terdiri dari in yang bernilai True apabila nilai ditemukan, dan not in yang bernilai True apabila nilai tidak ditemukan. Sementara itu, operator identitas digunakan untuk membandingkan apakah dua variabel merujuk pada objek yang sama di dalam memori, bukan sekadar membandingkan nilainya, dan terdiri dari is serta is not. Dua variabel dapat memiliki nilai yang sama tetapi merupakan objek yang berbeda, sehingga hasil operator identitas dapat berbeda dengan hasil operator perbandingan nilai (==).

## 2.1.6 Operator Perbandingan<br>
Operator perbandingan digunakan untuk membandingkan dua buah nilai dan menghasilkan nilai boolean sebagai hasilnya. Operator ini terdiri dari lebih besar dari (>), lebih kecil dari (<), lebih besar dari atau sama dengan (>=), lebih kecil dari atau sama dengan (<=), sama dengan (==), dan tidak sama dengan (!=). Python juga mendukung perbandingan berantai, yaitu penulisan lebih dari satu operator perbandingan dalam satu ekspresi, seperti 0 < x < 100, yang akan diproses sebagai gabungan dari dua kondisi perbandingan sekaligus.

### 2.2.2 Program "Assignment_2611531017"

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_1017 = int(input("Input angka-1: "))
angka2_1017 = int(input("Input angka-2: "))

print("\nNilai awal angka1_1017 =", angka1_1017)
print("Nilai angka2_1017 =", angka2_1017)

# Assignment biasa
hasil_1017 = angka1_1017
print("\nAssignment biasa (=)")
print("Hasil =", hasil_1017)

# Assignment penambahan
hasil_1017 = angka1_1017
hasil_1017 += angka2_1017
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_1017)

# Assignment pengurangan
hasil_1017 = angka1_1017
hasil_1017 -= angka2_1017
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_1017)

# Assignment perkalian
hasil_1017 = angka1_1017
hasil_1017 *= angka2_1017
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_1017)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_1017 != 0:
    hasil_1017 = angka1_1017
    hasil_1017 /= angka2_1017
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_1017)
    # Operator tambahan
    hasil_1017 = angka1_1017
    hasil_1017 //= angka2_1017
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_1017)
    hasil_1017 = angka1_1017
    hasil_1017 %= angka2_1017
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_1017)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_1017 = angka1_1017
hasil_1017 **= angka2_1017
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_1017)
```

### 2.2.3 Program "Bitwise_2611531017"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_1017 = int(input("Masukkan angka bitwise-1: "))
angka2_1017 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_1017, "| biner =", bin(angka1_1017))
print("angka2 =", angka2_1017, "| biner =", bin(angka2_1017))

# Bitwise AND
hasil_1017 = angka1_1017 & angka2_1017
print("\nBitwise AND (&)")
print(angka1_1017, "&", angka2_1017, "=", hasil_1017)
print("Biner hasil =", bin(hasil_1017))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))

# Bitwise OR
hasil_1017 = angka1_1017 | angka2_1017
print("\nBitwise OR (|)")
print(angka1_1017, "|", angka2_1017, "=", hasil_1017)
print("Biner hasil =", bin(hasil_1017))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))

# Bitwise XOR
hasil_1017 = angka1_1017 ^ angka2_1017
print("\nBitwise XOR (^)")
print(angka1_1017, "^", angka2_1017, "=", hasil_1017)
print("Biner hasil =", bin(hasil_1017))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))

# Bitwise NOT
hasil_1017 = ~angka1_1017
print("\nBitwise NOT (~)")
print("~", angka1_1017, "=", hasil_1017)
print("Biner hasil =", bin(hasil_1017))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))

# Bitwise geser kiri
jumlah_geser_1017 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1017 = angka1_1017 << jumlah_geser_1017
print("\nBitwise geser kiri (<<)")
print(angka1_1017, "<<", jumlah_geser_1017, "=", hasil_1017)
print("Biner hasil =", bin(hasil_1017))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))

# Bitwise geser kanan
hasil_1017 = angka1_1017 >> jumlah_geser_1017
print("\nBitwise geser kanan (>>)")
print(angka1_1017, ">>", jumlah_geser_1017, "=", hasil_1017)
print("Biner hasil =", bin(hasil_1017))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))
```
