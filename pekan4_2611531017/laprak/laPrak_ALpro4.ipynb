## 2.1 Dasar Teori <br>
Pada praktikum ini dibahas beberapa bentuk struktur kondisional dalam bahasa pemrograman Python, yaitu if, if-else, if-elif-else, multi-if, serta match-case. Struktur-struktur tersebut digunakan agar program dapat mengambil keputusan dan menjalankan blok kode tertentu berdasarkan syarat yang diberikan, sehingga keluaran program dapat menyesuaikan dengan kondisi yang terjadi saat program dijalankan. Dasar teori dari masing-masing struktur kondisional yang dipelajari adalah sebagai berikut:

### 2.1.1 If<br>
Pernyataan if merupakan struktur kondisional paling dasar dalam Python yang digunakan untuk menjalankan suatu blok kode apabila syarat yang diberikan bernilai benar. Penulisannya diawali dengan kata kunci if yang diikuti oleh suatu ekspresi kondisi dan tanda titik dua, kemudian blok kode yang akan dijalankan ditulis pada baris berikutnya dengan indentasi. Apabila ekspresi kondisi bernilai benar, blok kode di dalamnya akan dieksekusi, sedangkan apabila bernilai salah, blok kode tersebut akan dilewati begitu saja tanpa ada blok pengganti yang dijalankan.

### 2.1.2 If-Else<br>
Pernyataan if-else merupakan pengembangan dari pernyataan if yang digunakan apabila terdapat dua kemungkinan aksi yang perlu dijalankan berdasarkan satu syarat. Apabila ekspresi kondisi pada bagian if bernilai benar, maka blok kode pada bagian if yang akan dijalankan, sedangkan apabila bernilai salah, maka blok kode pada bagian else yang akan dijalankan sebagai gantinya, sehingga salah satu dari kedua blok kode tersebut pasti akan dieksekusi.

### 2.1.3 If-Elif-Else<br>
Pernyataan if-elif-else digunakan apabila terdapat lebih dari dua kemungkinan kondisi yang perlu diperiksa secara berurutan. Kata kunci elif merupakan singkatan dari "else if" yang berfungsi menambahkan syarat lanjutan setelah kondisi if sebelumnya bernilai salah. Python akan memeriksa setiap kondisi secara berurutan dari atas ke bawah, dan hanya blok kode pada kondisi pertama yang bernilai benar yang akan dieksekusi, sedangkan kondisi-kondisi setelahnya tidak lagi diperiksa. Jumlah klausa elif dapat lebih dari satu, sementara klausa else di bagian akhir bersifat opsional dan hanya dijalankan apabila seluruh kondisi sebelumnya bernilai salah.

### 2.1.4 Multi-If<br>
Multi-if merupakan istilah untuk penggunaan beberapa pernyataan if secara terpisah dan berdiri sendiri-sendiri, bukan dirangkai menggunakan elif. Berbeda dengan if-elif-else yang hanya menjalankan satu blok kode dari kondisi pertama yang benar, pada multi-if setiap pernyataan if diperiksa secara independen tanpa memandang hasil dari pernyataan if lainnya. Bentuk ini cocok digunakan ketika syarat-syarat yang diperiksa bersifat tidak saling berkaitan (non-exclusive), sehingga apabila lebih dari satu kondisi bernilai benar secara bersamaan, lebih dari satu blok kode juga dapat dieksekusi sekaligus.

### 2.1.5 Match-Case<br>
Match-case merupakan struktur kondisional yang diperkenalkan pada Python versi 3.10 sebagai bentuk pencocokan pola atau pattern matching. Struktur ini menggunakan kata kunci match yang diikuti oleh suatu nilai subjek, kemudian nilai tersebut dibandingkan dengan sejumlah pola yang dituliskan pada masing-masing blok case secara berurutan dari atas ke bawah. Hanya blok case pertama yang cocok dengan nilai subjek yang akan dieksekusi, sedangkan tanda garis bawah (_) dapat digunakan sebagai pola wildcard yang akan selalu cocok apabila tidak ada pola lain yang sesuai. Struktur ini sering dijadikan alternatif dari if-elif-else yang bertingkat banyak, terutama ketika suatu nilai perlu dibandingkan dengan beberapa kemungkinan nilai konstan secara lebih ringkas dan terstruktur.

### 2.2.5 Program "multi_if2_2611531017"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_1017 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1017 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_1017 = input_member_1017 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_1017 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1017 = input_promo_1017 in ["y", "ya"]

total_diskon_persen_1017 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1017 > 1000000:
    total_diskon_persen_1017 += 10  # Diskon belanja besar

if is_member_1017:
    total_diskon_persen_1017 += 5   # Diskon member

if kode_promo_valid_1017:
    total_diskon_persen_1017 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1017 = total_belanja_1017 * (total_diskon_persen_1017 / 100)
total_bayar_1017 = total_belanja_1017 - nominal_diskon_1017

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_1017}% (Rp {nominal_diskon_1017:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_1017:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1017}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

```python
# =========================================================
# Program: Sistem Loket Terpadu & Audit Transaksi
#          Ekspedisi Wahana - Alpro Adventure Park
# Materi : if, if-else, if-elif-else, multi-if, match-case
# Semua variabel diakhiri NIM: 1017
# =========================================================

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# ---------------------------------------------------------
# 1. Input Data Pengunjung
# ---------------------------------------------------------
nama_1017 = input("Masukkan Nama Pengunjung        : ")
umur_1017 = int(input("Input umur anda                 : "))
sim_1017 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

# ---------------------------------------------------------
# 2. Pemilihan Wahana Menggunakan Match-Case
# ---------------------------------------------------------
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_1017 = int(input("Masukkan nomor paket (1-5)      : "))

match paket_1017:
    case 1:
        nama_wahana_1017 = "Wahana Safari Rimba"
        harga_satuan_1017 = 50000
    case 2:
        nama_wahana_1017 = "Wahana Arung Jeram"
        harga_satuan_1017 = 75000
    case 3:
        nama_wahana_1017 = "Wahana Motor ATV Ekstrim"
        harga_satuan_1017 = 120000
    case 4:
        nama_wahana_1017 = "Wahana Roller Coaster Kilat"
        harga_satuan_1017 = 100000
    case 5:
        nama_wahana_1017 = "Wahana All-Access VIP"
        harga_satuan_1017 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

# ---------------------------------------------------------
# 3. Input Jumlah Tiket & Member/Promo
# ---------------------------------------------------------
jumlah_tiket_1017 = int(input("Masukkan jumlah tiket           : "))

# --- If tunggal: validasi kelogisan jumlah tiket ---
if jumlah_tiket_1017 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

is_member_1017 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_1017 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# ---------------------------------------------------------
# 4. Validasi Izin Kendali Wahana (If-Elif-Else + Operator Logika)
# ---------------------------------------------------------
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_1017 == 3:
    if umur_1017 >= 17 and sim_1017 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_1017 >= 17 and sim_1017 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_1017 < 17 and sim_1017 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_1017 >= 10:
        print("Status Akses: Anda diperbolehkan menaiki wahana ini.")
    else:
        print("Status Akses: Anda belum cukup umur untuk menaiki wahana ini.")

# ---------------------------------------------------------
# 5. Akumulasi Diskon Bertingkat (Multi-If Terpisah)
# ---------------------------------------------------------
subtotal_1017 = harga_satuan_1017 * jumlah_tiket_1017
total_diskon_persen_1017 = 0

if subtotal_1017 >= 200000:
    total_diskon_persen_1017 += 10  # Diskon Belanja Besar

if is_member_1017 in ['y', 'ya']:
    total_diskon_persen_1017 += 5   # Diskon Member

if kode_promo_valid_1017 in ['y', 'ya']:
    total_diskon_persen_1017 += 15  # Diskon Voucher Promo

if jumlah_tiket_1017 >= 5:
    total_diskon_persen_1017 += 5   # Diskon Tambahan Rombongan

# ---------------------------------------------------------
# 6. Evaluasi Kelulusan Audit (If-Else)
# ---------------------------------------------------------
nominal_diskon_1017 = subtotal_1017 * (total_diskon_persen_1017 / 100)
total_bayar_1017 = subtotal_1017 - nominal_diskon_1017

if total_bayar_1017 > 300000:
    catatan_layanan_1017 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    
# ---------------------------------------------------------
# 7. Cetak Rincian Pembayaran
# ---------------------------------------------------------
print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_1017:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_1017}% (Rp {nominal_diskon_1017:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_1017:,.0f}")
if total_bayar_1017 > 300000:
    print("Selamat! Anda berhak mendapatkan Souvenir Gratis.")
print(" Catatan Layanan  : Terima Kasih Telah Berkunjung")
print("Program Selesai")
```
