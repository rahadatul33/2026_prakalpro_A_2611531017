## 2.1 Dasar Teori

Pada praktikum ini dibahas beberapa konsep dasar dalam pemrograman Python, yaitu perulangan `for`, fungsi `range()`, nested loop, serta penggunaan parameter `end` pada fungsi `print()`. Keempat konsep tersebut digunakan untuk mengatur proses perulangan dan keluaran program secara terstruktur. Dasar teori dari masing-masing konsep adalah sebagai berikut.

### 2.1.1 Perulangan For
Perulangan for merupakan salah satu struktur perulangan pada Python yang digunakan untuk mengulangi suatu proses terhadap setiap elemen dalam sebuah urutan (sequence), seperti list, string, atau objek yang dapat diiterasi. Berbeda dengan beberapa bahasa pemrograman lain yang umumnya menggunakan kondisi atau pencacah secara langsung, for pada Python bekerja dengan mengambil setiap elemen dari suatu objek secara berurutan.

Perulangan for digunakan ketika suatu instruksi perlu dilakukan berulang kali dengan pola yang sama. Setiap kali perulangan berlangsung, variabel perulangan akan menerima nilai dari elemen yang sedang diproses. Setelah seluruh elemen selesai diproses, perulangan akan berhenti secara otomatis. Dengan demikian, for dapat membantu membuat program lebih sederhana dan mengurangi penulisan instruksi yang sama secara berulang.

### 2.1.2 Fungsi range()
Fungsi range() merupakan fungsi bawaan Python yang digunakan untuk menghasilkan suatu urutan bilangan, terutama ketika digunakan bersama perulangan for. Fungsi ini dapat menentukan batas awal dan batas akhir dari bilangan yang akan dihasilkan. Secara umum, range() dapat digunakan dengan menentukan nilai akhir saja atau dengan menentukan nilai awal dan nilai akhir.

Hal penting dalam penggunaan range() adalah nilai batas akhir (stop) tidak termasuk ke dalam urutan yang dihasilkan. Oleh karena itu, fungsi ini dapat digunakan untuk mengatur berapa kali suatu perulangan dijalankan. Selain itu, range() juga dapat memiliki nilai langkah (step) untuk menentukan jarak antarbilangan. Penggunaan range() membuat jumlah dan urutan iterasi dalam suatu perulangan dapat ditentukan dengan lebih teratur.

### 2.1.3 Nested Loop
Nested loop atau perulangan bersarang adalah perulangan yang terdapat di dalam perulangan lainnya. Pada nested loop, perulangan yang berada di luar disebut *outer loop*, sedangkan perulangan yang berada di dalam disebut *inner loop*. Setiap kali satu iterasi pada *outer loop* berlangsung, *inner loop* akan dijalankan sampai selesai.

Nested loop banyak digunakan ketika program membutuhkan pengulangan dalam lebih dari satu tingkat, misalnya untuk membentuk pola berdasarkan baris dan kolom. Dalam pembuatan pola, *outer loop* biasanya digunakan untuk mengatur jumlah baris, sedangkan *inner loop* digunakan untuk mengatur isi atau jumlah karakter pada setiap baris. Dengan konsep ini, program dapat menghasilkan keluaran yang memiliki susunan tertentu secara sistematis.

### 2.1.4 Parameter end pada Fungsi print()
Fungsi print() pada Python digunakan untuk menampilkan informasi atau objek ke keluaran program. Secara bawaan, fungsi print() akan mengakhiri hasil keluaran dengan karakter baris baru (\n), sehingga keluaran berikutnya akan ditampilkan pada baris selanjutnya.

Parameter end digunakan untuk menentukan karakter atau teks yang ditempatkan setelah objek yang dicetak. Nilai bawaan parameter end adalah karakter baris baru. Dengan mengubah nilai end, keluaran dapat tetap berada pada baris yang sama atau diakhiri dengan karakter tertentu. Penggunaan parameter end sangat berguna dalam pembuatan pola menggunakan nested loop, karena karakter yang dicetak oleh *inner loop* dapat ditempatkan secara berurutan pada satu baris sebelum program berpindah ke baris berikutnya.

### 2.2.7 Program "nested_for4_NIM"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_1017 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1017 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1017 = tinggi_1017
    c_1017 = a_1017
    lebar_1017 = (2 * tinggi_1017) - 2

    for i_1017 in range(1, tinggi_1017 + 1):
        b_1017 = c_1017 + 1

        for j_1017 in range(1, lebar_1017 + 1):

            # Baris atas dan bawah
            if i_1017 == 1 or i_1017 == tinggi_1017:
                if j_1017 == 1 or j_1017 == lebar_1017:
                    print("#", end="")
                else:
                    print("=", end="")

                    # Baris isi
            else:
                if j_1017 == 1 or j_1017 == lebar_1017:
                    print("|", end="")
                else:
                    if j_1017 == c_1017:
                        print("<", end="")
                    elif j_1017 == b_1017:
                        print(">", end="")
                    elif j_1017 == (lebar_1017 - c_1017):
                        print("<", end="")
                    elif j_1017 == (lebar_1017 - c_1017 + 1):
                        print(">", end="")
                    elif j_1017 > b_1017 and j_1017 < (lebar_1017 - c_1017):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1017 -= 2

        if a_1017 <= 0:
            c_1017 = (-a_1017) + 2
        else:
            c_1017 = a_1017
```

**Penjelasan:**

1. Baris 1–4 berfungsi sebagai komentar yang menjelaskan nama file, tujuan program, aturan penamaan variabel, dan penggunaan fungsi input().

```python
# Buat file dengan nama tugas5_nim.py
# Program Jam Pasir Kristal Palindromik Berbingkai menggunakan for bersarang
# Nama variabel ditambah 4 digit nim terakhir contoh: n_1234
# Program ini menggunakan fungsi input()

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_1017 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Bingkai atas
print("#", end="")
for garis_1017 in range(4 * n_1017 + 5):
    print("=", end="")
print("#", end="")
print()

# Fase 1: jam pasir atas (baris N turun sampai 1)
for i_1017 in range(1, n_1017 + 1):
    baris_1017 = n_1017 - i_1017 + 1
    print("|", end="")
    print(" ", end="")
    for kiri_1017 in range(2 * (n_1017 - baris_1017)):
        print(" ", end="")
    for mundur_1017 in range(1, baris_1017 + 1):
        print(baris_1017 - mundur_1017 + 1, end="")
        print(" ", end="")
    print("<*>", end="")
    for maju_1017 in range(1, baris_1017 + 1):
        print(" ", end="")
        print(maju_1017, end="")
    for kanan_1017 in range(2 * (n_1017 - baris_1017)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# Fase 2: poros titik pusat
print("|", end="")
for kiri_1017 in range(2 * n_1017 + 1):
    print(" ", end="")
print("<*>", end="")
for kanan_1017 in range(2 * n_1017 + 1):
    print(" ", end="")
print("|", end="")
print()

# Fase 3: jam pasir bawah (baris 1 naik sampai N)
for baris_1017 in range(1, n_1017 + 1):
    print("|", end="")
    print(" ", end="")
    for kiri_1017 in range(2 * (n_1017 - baris_1017)):
        print(" ", end="")
    for mundur_1017 in range(1, baris_1017 + 1):
        print(baris_1017 - mundur_1017 + 1, end="")
        print(" ", end="")
    print("<*>", end="")
    for maju_1017 in range(1, baris_1017 + 1):
        print(" ", end="")
        print(maju_1017, end="")
    for kanan_1017 in range(2 * (n_1017 - baris_1017)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# Bingkai bawah
print("#", end="")
for garis_1017 in range(4 * n_1017 + 5):
    print("=", end="")
print("#", end="")
print()
```
